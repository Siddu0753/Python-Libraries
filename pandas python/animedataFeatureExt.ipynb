```python
def extract_episodes(txt):
    check=False
    data=""
    for i in txt:
        if i==')':
            check=False
            return data
        if check==True:
            data+=i
        if i=='(':
            check=True
    
            
```

```python
df['episodes']=df['Title'].apply(extract_episodes)
```

```python
df['episodes']=df['episodes'].str.replace("eps","")
```

```python
df['episodes']=df['episodes'].astype(int)
```

```python
#make a new column for timestamp
```

```python
def extraction_time(txt):
    check=False
    data=""
    for i in range(len(txt)):
        if txt[i]==')':
            for j in range(i+1,i+20):
                data+=txt[j]

            return data
                
            
        
```

```python
df['total time']=df['Title'].apply(extraction_time)
```

```python
from datetime import datetime
```

```python
import pandas as pd

# Convert each duration to months
def get_total_months(duration):
    start, end = duration.split(" - ")

    start = pd.to_datetime(start, format="%b %Y")
    end = pd.to_datetime(end, format="%b %Y")

    months = (end.year - start.year) * 12 + (end.month - start.month)
    return months

# Create a new column
df["total months"] = df["total time"].apply(get_total_months)

```
